```python
def greenComp(image):
        if image is None or image.ndim < 3 or image.shape[2] < 2:
            print("Invalid image input for greenComp.")
            return np.array([])

        gcImg = image[:, :, 1]  # Extract the green channel
        return gcImg
```

```python
def plot_img(image):
    plt.figure(figsize=(12, 5))
    plt.imshow(image)
```

```python
image_path = "Screenshot 2024-11-20 184405.png"
original_image = cv.imread(image_path)
image = cv.resize(original_image, (224, 224))
plot_img(image)
img = greenComp(image)
```

```python
def histEqualize(image):
        if image.size == 0:
            print("Input image for histEqualize is empty.")
            return np.array([])

        equalized_image = cv.equalizeHist(image)
        return equalized_image
```

```python
import numpy as np

def histEq(img):
    hist,bins = np.histogram(img.flatten(),256,[0,256])
    cdf = hist.cumsum()
    cdf_normalized = cdf * float(hist.max()) / cdf.max()
    
    plt.plot(cdf_normalized, color = 'b')
    plt.hist(img.flatten(),256,[0,256], color = 'r')
    plt.xlim([0,256])
    plt.legend(('cdf','histogram'), loc = 'upper left')
    plt.show()

    cdf_m = np.ma.masked_equal(cdf,0)
    cdf_m = (cdf_m - cdf_m.min())*255/(cdf_m.max()-cdf_m.min())
    cdf = np.ma.filled(cdf_m,0).astype('uint8')
    plt.figure(figsize = (12, 5))
    # plt.imshow(cdf[img], cv.COLOR_BGR2RGB)
    plt.imshow(cv.cvtColor(cdf[img], cv.COLOR_BGR2RGB))
    plt.show()
    print("\n\n\n\n")

    
    plt.plot(cdf, color = 'b')
    plt.hist(img.flatten(),256,[0,256], color = 'r')
    plt.xlim([0,256])
    plt.legend(('cdf','histogram'), loc = 'upper left')
    plt.show()    
    img2 = cdf[img]
    plt.imshow(img2)
```

```python
def grayScaling(img):
    features = cv.resize(img, (224, 224))
    features = features.astype('float32') / 255.0
    features = np.expand_dims(features, axis=-1)  # Add channel dimension
    return features
```

```python
def plot_img2(img):
    plt.figure(figsize = (15, 5))
    plt.subplot(1, 3, 2)
    plt.title("Extracted Blood Vessels")
    plt.imshow(features.squeeze(), cmap='gray')
    plt.axis('off')
```
